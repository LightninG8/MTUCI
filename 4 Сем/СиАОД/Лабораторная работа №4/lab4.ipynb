```python
import random
import time

def generate_random_data(size, min_val=0, max_val=10000000):
    return sorted(random.sample(range(min_val, max_val), size))

def time_execution(func, *args):
    start_time = time.time()
    result = func(*args)
    execution_time = time.time() - start_time
    return result, execution_time


# Бинарный поиск
def binary_search(data, target):
    left, right = 0, len(data) - 1
    while left <= right:
        mid = (left + right) // 2
        if data[mid] == target:
            return mid
        elif data[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1

def add_binary_search(data, value):
    data.append(value)
    data.sort()

def remove_binary_search(data, value):
    if value in data:
        data.remove(value)


# Бинарное дерево
class TreeNode:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

class BinarySearchTree:
    def __init__(self):
        self.root = None

    def insert(self, value):
        if not self.root:
            self.root = TreeNode(value)
        else:
            self._insert(self.root, value)

    def _insert(self, node, value):
        if value < node.value:
            if not node.left:
                node.left = TreeNode(value)
            else:
                self._insert(node.left, value)
        elif value > node.value:
            if not node.right:
                node.right = TreeNode(value)
            else:
                self._insert(node.right, value)

    def search(self, value):
        return self._search(self.root, value)

    def _search(self, node, value):
        if not node:
            return False
        if node.value == value:
            return True
        if value < node.value:
            return self._search(node.left, value)
        return self._search(node.right, value)

    def delete(self, value):
        self.root = self._delete(self.root, value)

    def _delete(self, node, value):
        if not node:
            return None
        if value < node.value:
            node.left = self._delete(node.left, value)
        elif value > node.value:
            node.right = self._delete(node.right, value)
        else:
            if not node.left and not node.right:
                return None
            if not node.left:
                return node.right
            if not node.right:
                return node.left
            min_larger_node = self._get_min(node.right)
            node.value = min_larger_node.value
            node.right = self._delete(node.right, min_larger_node.value)
        return node

    def _get_min(self, node):
        while node.left:
            node = node.left
        return node


# Фибоначчиев поиск
def fibonacci_search(data, target):
    fib2 = 0
    fib1 = 1
    fib = fib1 + fib2
    n = len(data)
    
    while fib < n:
        fib2 = fib1
        fib1 = fib
        fib = fib1 + fib2
    offset = -1
    
    while fib > 1:
        i = min(offset + fib2, n - 1)
        if data[i] < target:
            fib = fib1
            fib1 = fib2
            fib2 = fib - fib1
            offset = i
        elif data[i] > target:
            fib = fib2
            fib1 -= fib2
            fib2 = fib - fib1
        else:
            return i
        
    if fib1 and offset + 1 < n and data[offset + 1] == target:
        return offset + 1
    return -1

def add_fibonacci(data, value):
    data.append(value)
    data.sort()

def remove_fibonacci(data, value):
    if value in data:
        data.remove(value)


# Интерполяционный поиск
def interpolation_search(data, target):
    low = 0
    high = len(data) - 1

    while low <= high and target >= data[low] and target <= data[high]:
        if data[high] == data[low]:
            if data[low] == target:
                return low
            return -1
        pos = low + ((high - low) * (target - data[low]) // (data[high] - data[low]))
        if pos < 0 or pos >= len(data):
            return -1
        if data[pos] == target:
            return pos
        elif data[pos] < target:
            low = pos + 1
        else:
            high = pos - 1
    return -1


def add_interpolation(data, value):
    data.append(value)
    data.sort()

def remove_interpolation(data, value):
    if value in data:
        data.remove(value)

def test_searches(data, bst):
    target = random.choice(data)

    _, native_time = time_execution(lambda d, t: t in d, data, target)

    bin_index, bin_time = time_execution(binary_search, data, target)
    bst_result, bst_time = time_execution(bst.search, target)
    fib_index, fib_time = time_execution(fibonacci_search, data, target)
    interp_index, interp_time = time_execution(interpolation_search, data, target)

    print("------------------------------------------------------------")
    print(f"Искомый элемент: {target}")
    print(f"Стандартный поиск: {'найден' if target in data else 'не найден'}, время: {native_time:.10f} сек")
    print(f"Бинарный поиск: {'найден' if bin_index != -1 else 'не найден'}, время: {bin_time:.10f} сек")
    print(f"Бинарное дерево: {'найден' if bst_result else 'не найден'}, время: {bst_time:.10f} сек")
    print(f"Фибоначчиев поиск: {'найден' if fib_index != -1 else 'не найден'}, время: {fib_time:.10f} сек")
    print(f"Интерполяционный поиск: {'найден' if interp_index != -1 else 'не найден'}, время: {interp_time:.10f} сек")
    print("------------------------------------------------------------")


data_size = 1000
data = generate_random_data(data_size)

bst = BinarySearchTree()
for value in data:
    bst.insert(value)

test_searches(data, bst)

new_element = 123456
add_binary_search(data, new_element)
bst.insert(new_element)
add_fibonacci(data, new_element)
add_interpolation(data, new_element)

print(f"Добавлен новый элемент: {new_element}")
test_searches(data, bst)

remove_binary_search(data, new_element)
bst.delete(new_element)
remove_fibonacci(data, new_element)
remove_interpolation(data, new_element)

print(f"Удалён элемент: {new_element}")
test_searches(data, bst)
```

### Задание №2:

Простое рехэширование | Рехэширование с помощью псевдослучайных чисел | Метод цепочек



### Выполнение:

```python
import random
import time

def generate_random_data(size, min_val=0, max_val=1000000):
    return random.sample(range(min_val, max_val), size)

def time_execution(func, *args):
    start_time = time.time()
    result = func(*args)
    execution_time = time.time() - start_time
    return result, execution_time

# Простое рехэширование
class LinearProbingHashTable:
    def __init__(self, size):
        self.size = size
        self.table = [None] * size
        self.deleted = [False] * size

    def add(self, key):
        index = key % self.size
        initial_index = index
        while self.table[index] is not None and not self.deleted[index]:
            index = (index + 1) % self.size
            if index == initial_index:
                raise Exception("Хэш-таблица переполнена!")
        self.table[index] = key
        self.deleted[index] = False

    def search(self, key):
        index = key % self.size
        initial_index = index
        while self.table[index] is not None:
            if self.table[index] == key and not self.deleted[index]:
                return index
            index = (index + 1) % self.size
            if index == initial_index:
                break
        return -1

    def delete(self, key):
        index = self.search(key)
        if index != -1:
            self.deleted[index] = True

# Рехэширование с псевдослучайными числами
class RandomProbingHashTable:
    def __init__(self, size):
        self.size = size
        self.table = [None] * size
        self.deleted = [False] * size
        self.random_offsets = [random.randint(1, size - 1) for _ in range(size)]

    def add(self, key):
        index = key % self.size
        initial_index = index
        i = 0
        while self.table[index] is not None and not self.deleted[index]:
            index = (index + self.random_offsets[i]) % self.size
            i += 1
            if index == initial_index:
                raise Exception("Хэш-таблица переполнена!")
        self.table[index] = key
        self.deleted[index] = False

    def search(self, key):
        index = key % self.size
        initial_index = index
        i = 0
        while self.table[index] is not None:
            if self.table[index] == key and not self.deleted[index]:
                return index
            index = (index + self.random_offsets[i]) % self.size
            i += 1
            if index == initial_index:
                break
        return -1

    def delete(self, key):
        index = self.search(key)
        if index != -1:
            self.deleted[index] = True

# Метод цепочек
class ChainingHashTable:
    def __init__(self, size):
        self.size = size
        self.table = [[] for _ in range(size)]

    def add(self, key):
        index = key % self.size
        if key not in self.table[index]:
            self.table[index].append(key)

    def search(self, key):
        index = key % self.size
        for stored_key in self.table[index]:
            if stored_key == key:
                return index
        return -1

    def delete(self, key):
        index = key % self.size
        if key in self.table[index]:
            self.table[index].remove(key)

def test_hash_methods(data, linear_table, random_table, chaining_table):
    target = random.choice(data + [random.randint(10000001, 20000000)])
    _, list_search_time = time_execution(lambda: target in data)
    _, linear_time = time_execution(linear_table.search, target)
    _, random_time = time_execution(random_table.search, target)
    _, chaining_time = time_execution(chaining_table.search, target)

    print("------------------------------------------------------------")
    print(f"Искомый элемент: {target}")
    print(f"Обычный поиск в списке: {'найден' if target in data else 'не найден'}, время: {list_search_time:.10f} сек")
    print(f"Простое рехэширование: {'найден' if linear_table.search(target) != -1 else 'не найден'}, время: {linear_time:.10f} сек")
    print(f"Рехэширование с псевдослучайными числами: {'найден' if random_table.search(target) != -1 else 'не найден'}, время: {random_time:.10f} сек")
    print(f"Метод цепочек: {'найден' if chaining_table.search(target) != -1 else 'не найден'}, время: {chaining_time:.10f} сек")
    print("------------------------------------------------------------")

DATA_SIZE = 5000
TABLE_SIZE = 10000 

data = generate_random_data(DATA_SIZE)

linear_table = LinearProbingHashTable(TABLE_SIZE)
random_table = RandomProbingHashTable(TABLE_SIZE)
chaining_table = ChainingHashTable(TABLE_SIZE)

for value in data:
    linear_table.add(value)
    random_table.add(value)
    chaining_table.add(value)

test_hash_methods(data, linear_table, random_table, chaining_table)

new_value = random.randint(10000001, 20000000)
print(f"Добавление нового элемента: {new_value}")
linear_table.add(new_value)
random_table.add(new_value)
chaining_table.add(new_value)

test_hash_methods(data + [new_value], linear_table, random_table, chaining_table)

print(f"Удаление элемента: {new_value}")
linear_table.delete(new_value)
random_table.delete(new_value)
chaining_table.delete(new_value)

test_hash_methods(data, linear_table, random_table, chaining_table)
```

### Задание №3:

Расставить на стандартной 64-клеточной шахматной доске 8 ферзей так, чтобы ни один из них не находился под боем другого. Подразумевается, что ферзь бьёт все клетки, расположенные по вертикалям, горизонталям и обеим диагоналям.

Написать программу,  которая находит хотя бы один способ решения задач.


### Выполнение:

```python
def is_safe(board, row, col, n):
    for i in range(row):
        if board[i][col] == 1:
            return False

    i, j = row, col
    while i >= 0 and j >= 0:
        if board[i][j] == 1:
            return False
        i -= 1
        j -= 1

    i, j = row, col
    while i >= 0 and j < n:
        if board[i][j] == 1:
            return False
        i -= 1
        j += 1

    return True


def solve_n_queens(board, row, n):
    if row >= n:
        return True

    for col in range(n):
        if is_safe(board, row, col, n):
            board[row][col] = 1

            if solve_n_queens(board, row + 1, n):
                return True

            board[row][col] = 0

    return False

def print_board(board, n):
    for row in board:
        print(" ".join("Q" if col == 1 else "." for col in row))
    print("\n")

n = 8

board = [[0 for _ in range(n)] for _ in range(n)]

if solve_n_queens(board, 0, n):
    print("Решение найдено:")
    print_board(board, n)
else:
    print("Решение не существует.")
```
